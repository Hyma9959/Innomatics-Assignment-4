```python
def validate_recharge():
  valid_plans=[199,299,399,599]
  while True:
    amount=int(input("Enter recharge amount"))
    if amount>=50 and amount in valid_plans:
      print("Recharge Successful")
      break
    else:
      print("Invalid recharge! Try again")
validate_recharge()
```

```python
def check_inventory(products):
  for item, stock in products.items():
    print(f"{item}:{'Reorder Alert' if stock<15 else 'Stock OK'}")
products={"Pen":10,"Book":25,"Marker":5,"Bag":20}
check_inventory(products)
```

```python
def process_result(marks):
  total=0
  for m in marks:
    total+=m
  avg=total/len(marks)
  return "Pass" if avg>=50 else "Fail"
marks=list(map(int,input("Enter marks separated by space:").split()))
print(process_result(marks))
```

```python
def calculate_fare(km,peak):
  fare=50+(12 *km)
  if peak.lower()=="yes":
    fare*=1.25
  return fare
while True:
  km=float(input("Enter distance in (km):"))
  peak=input("Peak hour? (yes/no):")
  print("Total Fare:",calculate_fare(km,peak))
  if input("Retry?(yes/no):").lower()!="yes":
    break
```

```python
def check_eligibility(attendance):
  present=0
  for day in attendance:
    if day=="P":
      present+=1
  percent=(present/len(attendance))*100
  return "Eligible " if percent>=75 else "Not Eligible"
attendance=input("Enter attendance(P/A separated by space):").split()
print(check_eligibility(attendance))
```

```python
def check_password(password):
  has_digit=has_special=False
  for ch in password:
    if ch.isdigit():
      has_digit=True
    if ch in "@#$":
      has_special=True
  if len(password)>=8 and has_digit and has_special:
    return "Strong Password"
  return "Weak Password"
pwd=input("Enter password:")
print(check_password(pwd))
```
